```python
import numpy as np

# ==========================================
# Q1: Sensor Readings Analysis
# ==========================================

# Original sensor readings in Celsius
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])
print("Original readings:", temperature)

# (a) Add 2°C using vectorization
temp_corrected = temperature + 2
print("a. Corrected (+2°C):", temp_corrected)

# (b) Convert to Fahrenheit: F = (9/5)*C + 32
temp_fahrenheit = (9 / 5) * temp_corrected + 32
print("b. Fahrenheit:", temp_fahrenheit)

# (c) Readings > 32°C using Boolean indexing
gt_32 = temp_corrected[temp_corrected > 32]
print("c. Readings > 32°C:", gt_32)

# (d) Count readings exceeding 32°C
count_gt_32 = np.sum(temp_corrected > 32)
print("d. Count > 32°C:", count_gt_32)

# (e) Efficiency Explanation:
# Vectorization executes operations in low-level C code using SIMD (Single Instruction, Multiple Data) CPU instructions.
# It operates on contiguous memory blocks without Python for-loop overhead, dynamic type checking, or object creation.
# Boolean indexing creates a binary mask in C to filter data rapidly without conditional branching in Python.
```

```python
# ==========================================
# Q2: Step Tracker Matrix Analysis
# ==========================================

# 4 users x 3 days
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# (a) Total steps
print("a. Total steps:", np.sum(steps))

# (b) Mean steps
print("b. Mean steps:", np.mean(steps))

# (c) Max and Min
print("c. Max steps:", np.max(steps), "| Min steps:", np.min(steps))

# (d) Total steps for each day (axis=0)
print("d. Daily totals (axis=0):", np.sum(steps, axis=0))

# (e) Total steps for each user (axis=1)
print("e. User totals (axis=1):", np.sum(steps, axis=1))

# (f) Max step position using argmax()
flat_max = np.argmax(steps)
user_idx, day_idx = np.unravel_index(flat_max, steps.shape)
print(f"f. Argmax flat index: {flat_max} -> User {user_idx + 1}, Day {day_idx + 1} (Row {user_idx}, Col {day_idx})")
```

```python
# ==========================================
# Q3: Slicing, Views vs Copies, Flattening
# ==========================================

# (a) Create original array
original = np.array([1, 2, 3, 4, 5, 6])

# (b) Slice index 1 to 4
subset = original[1:5]
print("b. Subset (index 1 to 4):", subset)

# (c) Modify subset[0] = 999
subset[0] = 999
print("c. Modified subset[0]=999:")
print("   Subset:", subset)
print("   Original:", original, "(Slicing creates a VIEW, so original is modified)")

# (d) Slice with .copy()
original = np.array([1, 2, 3, 4, 5, 6])
copied_subset = original[1:5].copy()
copied_subset[0] = 500
print("d. Modified copied_subset[0]=500:")
print("   Copied subset:", copied_subset)
print("   Original:", original, "(.copy() isolates memory, original is unchanged)")

# (e) Reshape 1 to 12 into 3x4 matrix
matrix = np.arange(1, 13).reshape(3, 4)
print("e. 3x4 Matrix:\n", matrix)

# (f) Slicing
print("f. First row:", matrix[0])
print("   Last row:", matrix[-1])
print("   Second column:", matrix[:, 1])
print("   Rows 1-2 and Cols 2-3:\n", matrix[1:3, 2:4])

# (g, h, i) flatten() vs ravel()
r_arr = matrix.ravel()
f_arr = matrix.flatten()

r_arr[0] = 777
print("h. Modified ravel[0]=777 -> Original matrix [0,0]:", matrix[0, 0], "(ravel returns a VIEW)")

matrix = np.arange(1, 13).reshape(3, 4)
f_arr = matrix.flatten()
f_arr[0] = 888
print("i. Modified flatten[0]=888 -> Original matrix [0,0]:", matrix[0, 0], "(flatten returns a COPY)")

# (j) Properties
print(f"j. Shape: {matrix.shape}, ndim: {matrix.ndim}, size: {matrix.size}, dtype: {matrix.dtype}")
```

```python
# ==========================================
# Q4: OLS Regression Model
# ==========================================

# X: [Sleep, Activity, Stress] for 4 users
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# (a) Shape and dimensions
print(f"a. Shape of X: {X.shape}, ndim: {X.ndim}")

# (b) Transpose X.T (rows become features, cols become users)
print("b. X.T:\n", X.T)

# (c) Product X.T @ X
XTX = X.T @ X
print("c. X.T @ X:\n", XTX)

# (d) Inverse of X.T @ X
XTX_inv = np.linalg.inv(XTX)
print("d. Inverse (X.T @ X)^(-1):\n", XTX_inv)

# (e) OLS Beta coefficients: beta = (X^T X)^(-1) X^T y
beta = XTX_inv @ X.T @ y
print("e. Beta coefficients [Sleep, Activity, Stress]:", beta)

# (f) Explanation of coefficients:
# beta[0] = -0.1592: Additional sleep slightly reduces assistance score required (-0.16 per hour).
# beta[1] = +0.1408: Higher activity slightly increases assistance score (+0.14 per unit).
# beta[2] = +10.0630: Stress is the primary driver (+10.06 increase in assistance score per stress unit).

# (g) Predict for new_user = [5, 40, 7]
new_user = np.array([5, 40, 7])
pred_score = new_user @ beta
print(f"g. Predicted assistance score for new user {new_user}: {pred_score:.4f}")
```
